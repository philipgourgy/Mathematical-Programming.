### Problem 1: Triangle

Consider a triangle with vertices at $(0, 0)$, $(1, 0)$ and $(0, 1)$. Write a program that asks the user for a point $(x,y)$ and checks if the point is outside of the triangle, on the border of the triangle, or inside of the triangle. Some examples of expected behaviour:

    The x-coordinate: 0.5
    The y-coordinate: 0.5
    This point is on the border of the triangle.
---

    The x-coordinate: 0.25
    The y-coordinate: 0.25
    This point is on the inside of the triangle.
---

    The x-coordinate: 5
    The y-coordinate: 5
    This point is on the outside of the triangle.

```python
# Your solution goes here
x = float(input("The x-coordinate: "))
y = float(input("The y-coordinate: "))

if x + y == 1: 
    print("This point is on the border of the triangle")
elif x + y < 1: 
    print("This point is inside of the triangle")
else: 
    print("This point is outside the triangle")
```

### Problem 2: Quadratic equation

Write a program for solving a quadratic equation of the form $ax^2 + bx + c = 0$. The program asks the user for values $a$, $b$ and $c$. It should then use the quadratic formula to solve the equation:
$$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}.$$

Your program should check if $a$ is zero, as this would make the equation no longer quadratic. If $a = 0$, the program should display an error message: "The value of a cannot be zero for a quadratic equation". 

Your program should distinguish between the number of solutions there are to the quadratic equation. Examples of expected behaviour:

    Value of a: 1
    Value of b: 2
    Value of c: -8
    The roots are 2.0 and -4.0
---

    Value of a: 1
    Value of b: -2
    Value of c: 1
    The root is 1.0
---

    Value of a: 1
    Value of b: 0
    Value of c: 1
    There are no real roots
---

    Value of a: 0
    Value of b: 1
    Value of c: 1
    The value of a cannot be zero for a quadratic equation

```python
# Your solution goes here
import math
while True: 
    a = float(input("Value of a: "))
    b = float(input("Value of b: "))
    c = float(input("Value of c: "))
    if a == 0: 
        print("The value of a cannot be zero for a quadratic equation")
        break

    D = b**2 - (4*a*c)
    if D > 0: 
        root1 = (-b + math.sqrt(D))/2*a
        root2 = (-b - math.sqrt(D))/2*a
        print(f"The roots are {root1} and {root2}")
        break 
    elif D == 0:
        root = (-b + math.sqrt(D))/2*a
        print(f"The root is {root}")
        break
    else: 
        print("There are no real roots.")
        break
```

### Problem 3: Repeat password

Write a program that asks the user to input a password. The program should then ask the user to type the password again. ### Problem 5: Working with numbers

Write a program which asks the user for integer numbers. The program should keep asking for numbers until the user types in zero.

    Please type in integer numbers. Type in 0 to finish.
    Number: 5
    Number: 22
    Number: 9
    Number: -2
    Number: 0

#### Part 1: Number of inputs
After reading in the numbers the program should print out how many numbers were types in. The zero at the end should not be included in the count. You will need a new variable here to keep track of the numbers typed in.

    ... the program asks for numbers
    Numbers typed in: 4

#### Part 2: The sum
The program should also print out the sum of all numbers typed in. The zero at the end should not be included in the calculation. The program should now print out the following:

    ... the program asks for numbers
    Numbers typed in: 4
    The sum of the numbers is 34

#### Part 3: The mean
The program should also print out the mean of the numbers. The zero at the end should not be included in the calculation. You may assume the user will always type in at least one valid non-zero numbers.

    ... the program asks for numbers
    Numbers typed in: 4
    The sum of the numbers is 34
    The mean of the numbers is 8.5

#### Part 4: Positive and negative inputs
The program should also print out statistics on how many numbers were positive and how many were negative. The zero at the end should not be included in your calculation.

    ... the program asks for numbers
    Numbers typed in: 4
    The sum of the numbers is 34
    The mean of the numbers is 8.5
    Positive numbers: 3
    Negative numbers: 1

```python
# Your solution goes here
number_inputs = 0 
sum = 0
mean = 0
n_pos = 0
n_neg = 0
while True: 
    n = int(input("Please type in integer numbers. Type in 0 to finish."))
    if n != 0:
        number_inputs += 1
        sum += n 
        if n > 0: 
            n_pos += 1 
        else: 
            n_neg += 1
    else:
        mean = sum / number_inputs
        print(f"Numbers typed in: {number_inputs}")
        print(f"The sum of the numbers is {sum}")
        print(f"The mean of the numbers is {mean}")
        print(f"Positve numbers: {n_pos}")
        print(f"Negative numbers: {n_neg}")
        break
```

### Problem 6: Collatz sequence

A *Collatz sequence* is a sequence that is defined as follows:

- The sequence begins with any positive integer, say $n$.
- If the integer $n$ is odd, then the next number in the sequence will be $3n+1$.
- If the integer $n$ is even, then the next number in the sequence will be $n/2$.
- Repeat this process until you encounter the integer 1.

Using this rule and starting with $n=13$, we generate the following sequence:

$$13 \to 40 \to 20 \to 10 \to 5 \to 16 \to 8 \to 4 \to 2 \to 1.$$

It can be seen that this sequence (starting at 13 and finishing at 1) contains 10 terms. Although it has not been proved yet, the Collatz conjecture posits that all starting numbers eventually reach 1.

Write a program that asks the user for a positive integer as input and prints the length of the generated Collatz sequence starting at this given value. The expected outcome should looks something like:

    Starting number of Collatz sequence: 13
    The length of the Collatz sequence: 10

```python
# Your solution goes here
n = int(input("Starting number of Collatz sequence: "))
l = 1
print(f"Starting number of Collatz sequence: {n}")
while n != 1:
    if n % 2 == 0: 
        n/= 2 
        l+=1
    else: 
        n = 3*n + 1 
        l+=1
print(f"The length of the Collatz sequence: {l}")
```

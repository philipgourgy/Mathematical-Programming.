### Problem 1: Greatest common divisor (GCD) using the Euclidean algorithm

The **greatest common divisor** (GCD) of two integers is the largest number that divides both of them without leaving a remainder. For example, the GVD of 48 and 18 is 6, because 6 is the largest number that divides both 48 and 18 evenly.

The **Euclidean algorithm** is a method for finding the GCD of two numbers through a series of division steps. It works by repeatedly dividing the larger number by the smaller one and replacing the larger number with the remainder, until the remainder is zero. The last non-zero remainder is the GCD.

Here's how the Euclidean algorithm works, step by step, illustrated with an example. Let's use the numbers 48 and 18 and find their GCD using the Euclidean algorithm. 

1. Divide 48 by 18:
       $$48 \div 18 = 2 (\text{quotient}), \text{ remainder} = 48 - 18 \cdot 2 = 12.$$
   So the remainder is 12. Now replace 48 with 18, and 18 with the remainder 12.
3. Divide 18 by 12:
        $$18 \div 12 = 1 (\text{quotient}), \text{ remainder} = 18 - 12 \cdot 1 = 6.$$
   So the remainder is 6. Now replace 18 with 12, and 12 with the remainder 6.
5. Divide 12 by 6:
       $$12 \div 6 = 0 (\text{quotient}), \text{ remainder} = 12 - 6 \cdot 2 = 0.$$
   So the remainder is 0. Since the remainder is 0, we stop here.

The last non-zero remainder is 6, which is the GCD of 48 and 18.

### Problem 1: Fixed point iteration

A loan of $A$ euros is repaid by making $n$ equal monthly payments of $M$ euros, starting a month after the loan is made. It can be shown that if the monthly interest rate is $r$, then our monthly payments satisfy
    $$Ar = M\left(1 - \frac{1}{(1+r)^n}\right).$$
A car loan of €$10,000$ was repaid in $60$ monthly payments of €$250$. Use the fixed point iteration scheme to approximate the monthly interest rate and verify your solution. 

```python
def fixed_point_iteration(A, M, n, r0, tolerance=1e-6, max_iterations=1000):
    r = r0
    for _ in range(max_iterations):
        r_new = (M * (1 - (1 + r) ** -n)) / A
        if abs(r_new - r) < tolerance:
            return r_new
        r = r_new 
    print("Fixed point iteration did not converge within the maximum number of iterations.")

A = 10000  
M = 250    
n = 60     
r0 = 0.01
r_approx = fixed_point_iteration(A, M, n, r0)
print(f"Approximated monthly interest rate: {r_approx*100:.6f}%")
```

#### Part 2 - Algorithm

Use the pseudocode as structure to write a function `bisection`. The function should take as input a function, along with an initial lower bound, an initial upper bound, and the number of iterations. The output should be the midpoint of the final subinterval.

```python
# Solution problem 2b
def bisection(f, a, b, num_iterations):
    """
    Perform the bisection method to find a root of the equation f(x) = 0
    in the interval [a, b].

    Parameters:
    f (function): The function for which we are finding the root.
    a (float): The initial lower bound of the interval.
    b (float): The initial upper bound of the interval.
    num_iterations (int): The number of iterations to run the method.

    Returns:
    float: The midpoint of the final subinterval.
    """

    if f(a) * f(b) >= 0:
        print("Bisection method fails.")
        return None

    for _ in range(num_iterations):
        m = (a + b) / 2
        fm = f(m)
        if fm == 0:
            print("Found exact solution:", m)
            return m
        if f(a) * fm < 0:
            b = m 
        else:
            a = m  
    return (a + b) / 2
```

#### Part 4 - Approximation error

The bisection method cuts the width of the interval down in half at every step. As such, the approximation error gets smaller at every step. In the example above we know the exact solution, so we can compute it to the value of the midpoint given at each iteration by the bisection method and calculate the absolute error:
$$\text{Absolute error = |Approximate root - Exact root|} = |m_n - x^*|.$$
By Theorem 7.1 we know that
$$|m_n - x^*| \leq \frac{b-a}{2^{n+1}}.$$
So the error decreases exponentially with the number of iterations. ### Problem 3: Newton's method

#### Part 1 - Algorithm
Write a function called `newton` which takes input parameters `f`, `x0`, `h` (with default value 0.001), `tol` (with default value 0.001) and `max_iter` (with default value 100). The function implements Newton's method to approximate a solution of $f(x) = 0$. In other words, it computes the values of the recursive sequence starting at $x_0$ and defined by
$$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}.$$
Use the central difference formula $f'(x_n) = \frac{f(x_n + h) - f(x_n - h)}{2h}$ with stepsize $h$ to approximate the derivative $f'(x_n)$. The desired result is that the method converges to an approximate root of $f(x)$, however there are several possibilities:

- The sequence reaches the desired tolerance $|f(x_n)| \leq tol$ and `newton` returns the value $x_n$.
- The number of iterations exceeds the maximum number of iterations `max_iter`, the functions prints the statement `Maximum iterations exceeded` and returns `None`.
- A zero derivative is computed $f'(x_n) = 0$, the function prints the statement `Zero derivative` and returns `None`.

```python
# Solution problem 3a
def newton(f, x0, h=0.001, tol=0.001, max_iter=100):
    """
    Approximate solution of f(x) = 0 by Newton's method.

    Parameters
    ----------
    f (function):   Function for which we are searching a root
    x0 (float):     Initial guess for a solution f(x) = 0
    h (float):      Positive stepsize for central difference formula
    tol (float):    Stopping criteria is abs(f(x)) < tol
    max_iter (int): Maximum number of iterations of Newton's method

    Returns
    -------
    xn (float):     Approximate root if abs(f(xn)) < tol.
                    If f'(xn) = 0 or number of iterations exceeds
                    max_iter, then return None.
    
    """
    
    xn = x0
    
    for _ in range(max_iter):
        f_prime = (f(xn + h) - f(xn - h)) / (2 * h)
        if f_prime == 0:
            print("Zero derivative")
            return None
        xn_next = xn - f(xn) / f_prime

        if abs(f(xn_next)) <= tol:
            return xn_next
        xn = xn_next
    print("Maximum iterations exceeded")
    return None
```

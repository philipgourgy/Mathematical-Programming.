N.B. As it turns out, `round` implements are type of rounding called bankers' rounding. In most cases, bankers' rounding behaves as we might expect, namely it rounds to the closest integer. However, bankers' rounding is special in how it deals with ties. In the event of a tie, for example `3.5`, which is equally close to `3` and `4`, bankers' rounding always rounds towards the closest **even** number. Therefore `3.5` rounds towards `4`, but `2.5` rounds towards `2`. 

Why would we want this? The main reason is that for large sets of numbers, bankers' rounding is less biased, so a series of additions and subtractions with rounded numbers more accurately represents the true total of the unrounded numbers.
